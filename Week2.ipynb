7. Buat rectangle dan circle pada bagian wajah dari image foto s anda saat
beraktifitas (bukan pasfoto).

```python
face = cv.imread('/content/drive/MyDrive/PCVK/Images/Foto1.jpg')

face_rect = face.copy()
face_circle = face.copy()

cv.rectangle(face_rect, (140,270), (215,355), (0,255,0), 3)
cv.rectangle(face_rect, (610,340), (690,440), (0,255,0), 3)
cv.rectangle(face_rect, (475,370), (540,450), (0,255,0), 3)
cv.rectangle(face_rect, (230,230), (270,280), (0,255,0), 3)
cv.rectangle(face_rect, (285,275), (330,330), (0,255,0), 3)
cv.rectangle(face_rect, (430,370), (470,430), (0,255,0), 3)
cv.rectangle(face_rect, (395,370), (430,435), (0,255,0), 3)
cv.rectangle(face_rect, (420,270), (460,320), (0,255,0), 3)
cv.rectangle(face_rect, (420,270), (460,320), (0,255,0), 3)
cv.rectangle(face_rect, (220,300), (270,370), (0,255,0), 3)
cv.rectangle(face_rect, (350,330), (395,390), (0,255,0), 3)

rects = [
    (140,270,215,355),
    (610,340,690,440),
    (475,370,540,450),
    (230,230,270,280),
    (285,275,330,330),
    (430,370,470,430),
    (395,370,430,435),
    (420,270,460,320),
    (220,300,270,370),
    (350,330,395,390)
]

# gambar circle di posisi yang sama
for (x1,y1,x2,y2) in rects:
    cx = (x1 + x2) // 2
    cy = (y1 + y2) // 2
    r = max(x2 - x1, y2 - y1) // 2
    cv.circle(face_circle, (cx,cy), r, (205,0,0), 3)

plt.imshow(cv.cvtColor(face_rect, cv.COLOR_BGR2RGB))
plt.show()

plt.imshow(cv.cvtColor(face_circle, cv.COLOR_BGR2RGB))
plt.show()
```

8. Buat rectangle pada bagian sudut bawah kiri channel B pada color space
RGB dari citra kitten/ lena/ mandrill/ male/ female/ couple/ sailboat/
peppers!

9. Lengkapi tulisan nama file pada file citra dari soal no.8. gunakan font,
ukuran font, dan warna font yang sesuai keinginan anda

```python
# Daftar file gambar
files = ["kitten01.jpg", "lena.jpg", "mandrill.jpg", "male.jpg",
         "female.jpg", "couple.jpg", "sailboat.jpg", "peppers.jpg"]

# Loop untuk setiap file
for fname in files:
    # Baca gambar
    img = cv.imread(f"/content/drive/MyDrive/PCVK/Images/{fname}")

    # Pisahkan menjadi channel B, G, R
    B, G, R = cv.split(img)

    # Dapatkan ukuran gambar
    tinggi, lebar = B.shape

    # Gambar persegi panjang di channel Biru (pojok kiri bawah)
    cv.rectangle(B, (0, tinggi-100), (250, tinggi), (255,), -1)

    # Gabungkan kembali channel
    gabungan = cv.merge([B, G, R])

    # Tambahkan teks (nama file) di dalam kotak
    cv.putText(gabungan, fname, (10, tinggi-20),
               cv.FONT_HERSHEY_SIMPLEX, 0.8, (0,0,255), 2, cv.LINE_AA)

    # Tampilkan hasil
    plt.figure(figsize=(6,4))
    plt.title(f"Hasil - {fname}")
    plt.imshow(cv.cvtColor(gabungan, cv.COLOR_BGR2RGB))
    plt.axis("off")
    plt.show()
```
